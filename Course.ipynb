```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.read_csv("data_OneHot_Application_mode.csv", sep=";") # ler o bd

encoder = OneHotEncoder(sparse_output=False) # cria o OneHotEncoder

course_encoded = encoder.fit_transform(df[["Course"]]) # fazendo as matrizes

nomes = encoder.get_feature_names_out(["Course"])

course_df = pd.DataFrame(
    course_encoded,
    columns=nomes,
    index=df.index
)

df = pd.concat([df.drop(columns=["Course"]), course_df],axis=1) #Removendo a coluna antiga e juntando as novas

print(df.head())
```

```python
df.to_csv("data_OneHot_Course.csv", sep=";", index=False)
```
