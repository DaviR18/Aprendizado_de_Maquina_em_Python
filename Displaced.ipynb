```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.read_csv("data_OneHot_Father's_occupation.csv", sep=";")
encoder = OneHotEncoder(sparse_output=False)

Displaced_encoded = encoder.fit_transform(df[["Displaced"]])

nomes = encoder.get_feature_names_out(["Displaced"])

Displaced_df = pd.DataFrame(
    Displaced_encoded,
    columns=nomes,
    index=df.index
)

df = pd.concat([df.drop(columns=["Displaced"]), Displaced_df],axis=1)

print(df.head())

```

```python
df.to_csv("data_OneHot_Displaced.csv", sep=";", index=False)
```
