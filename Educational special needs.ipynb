```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.read_csv("data_OneHot_Displaced.csv", sep=";")
encoder = OneHotEncoder(sparse_output=False)

Educational_special_needs_encoded = encoder.fit_transform(df[["Educational special needs"]])

nomes = encoder.get_feature_names_out(["Educational special needs"])

Educational_special_needs_df = pd.DataFrame(
    Educational_special_needs_encoded,
    columns=nomes,
    index=df.index
)

df = pd.concat([df.drop(columns=["Educational special needs"]), Educational_special_needs_df],axis=1)

print(df.head())
```

```python
df.to_csv("data_OneHot_Educational_special_needs.csv", sep=";", index=False)
```
