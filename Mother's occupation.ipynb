```python
import pandas as pd

df = pd.read_csv("data_OneHot_Daytime_eveningattendance.csv", sep=";")

df.columns = df.columns.str.strip()

df["Mother's occupation"] = df["Mother's occupation"].apply(
    lambda x: "Unskilled Workers" if x == 9
    else "Administrative staff" if x == 4
    else "Personal Services, Security and Safety Workers and Sellers" if x == 5
    else "Intermediate Level Technicians and Professions" if x == 3
    else "Specialists in Intellectual and Scientific Activities" if x == 2
    else "Skilled Workers in Industry, Construction and Craftsmen" if x == 7
    else "Student" if x == 0
    else "Representatives of the Legislative Power and Executive Bodies, Directors, Directors and Executive Managers" if x == 1
    else "Farmers and Skilled Workers in Agriculture, Fisheries and Forestry" if x == 6
    else "Other Situation" if x == 90
    else "Installation and Machine Operators and Assembly Workers" if x == 8
    else "Other"
)

print(df["Mother's occupation"].value_counts())
```

```python
df.to_csv("data_Categoriarara(Mother's_occupation).csv", sep=";", index=False)
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.read_csv("data_Categoriarara(Mother's_occupation).csv", sep=";") # ler o bd

encoder = OneHotEncoder(sparse_output=False) # cria o OneHotEncoder

mothers_occupation_encoded = encoder.fit_transform(df[["Mother's occupation"]]) # fazendo as matrizes

nomes = encoder.get_feature_names_out(["Mother's occupation"])

mothers_occupation_df = pd.DataFrame(
    mothers_occupation_encoded,
    columns=nomes,
    index=df.index
)

df = pd.concat([df.drop(columns=["Mother's occupation"]), mothers_occupation_df],axis=1) #Removendo a coluna antiga e juntando as novas

print(df.head())
```

```python
df.to_csv("data_OneHot_Mother's_occupation.csv", sep=";", index=False)
```
