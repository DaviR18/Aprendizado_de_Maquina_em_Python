```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.read_csv("data_OneHot_Course.csv", sep=";") # ler o bd

encoder = OneHotEncoder(sparse_output=False) # cria o OneHotEncoder

Daytime_eveningattendance_encoded = encoder.fit_transform(df[["Daytime/evening attendance"]]) # fazendo as matrizes

nomes = encoder.get_feature_names_out(["Daytime/evening attendance"])

Daytime_eveningattendance_df = pd.DataFrame(
    Daytime_eveningattendance_encoded,
    columns=nomes,
    index=df.index
)

df = pd.concat([df.drop(columns=["Daytime/evening attendance"]), Daytime_eveningattendance_df],axis=1) #Removendo a coluna antiga e juntando as novas

print(df.head())
```

```python
df.to_csv("data_OneHot_Daytime_eveningattendance.csv", sep=";", index=False)
```
