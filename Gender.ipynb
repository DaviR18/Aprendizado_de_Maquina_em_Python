```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.read_csv("data_OneHot_Tuition_fees_up_to_date.csv", sep=";")
encoder = OneHotEncoder(sparse_output=False)

Gender_encoded = encoder.fit_transform(df[["Gender"]])

nomes = encoder.get_feature_names_out(["Gender"])

Gender_df = pd.DataFrame(
    Gender_encoded,
    columns=nomes,
    index=df.index
)

df = pd.concat([df.drop(columns=["Gender"]), Gender_df],axis=1)
print(df.head())
```

```python
df.to_csv("data_OneHot_Gender.csv", sep=";", index=False)
```
