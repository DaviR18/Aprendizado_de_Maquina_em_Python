```python
import pandas as pd

df = pd.read_csv("data_OneHot_Marital_status.csv", sep=";")

df.columns = df.columns.str.strip()

df["Application mode"] = df["Application mode"].apply(
    lambda x: "1st phase - general contingent" if x == 1
    else "2nd phase - general contingent" if x == 17
    else "Over 23 years old" if x == 39
    else "Change of course" if x == 43
    else "Technological specialization diploma holders" if x == 44
    else "Holders of other higher courses" if x == 7
    else "3rd phase - general contingent" if x == 18
    else "Transfer" if x == 42
    else "Change of institution/course" if x == 51
    else "Other"
)

print(df["Application mode"].value_counts())
```

```python
df.to_csv("data_Categoriarara(Application_mode).csv", sep=";", index=False)
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.read_csv("data_Categoriarara(Application_mode).csv", sep=";") # ler o bd

encoder = OneHotEncoder(sparse_output=False) # cria o OneHotEncoder

application_mode_encoded = encoder.fit_transform(df[["Application mode"]]) # fazendo as matrizes

nomes = encoder.get_feature_names_out(["Application mode"])

application_mode_df = pd.DataFrame(
    application_mode_encoded,
    columns=nomes,
    index=df.index
)

df = pd.concat([df.drop(columns=["Application mode"]), application_mode_df],axis=1) #Removendo a coluna antiga e juntando as novas

print(df.head())
```

```python
df.to_csv("data_OneHot_Application_mode.csv",sep=";",index=False)
```
