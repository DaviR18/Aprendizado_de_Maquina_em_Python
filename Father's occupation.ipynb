```python
import pandas as pd

df = pd.read_csv("data_OneHot_Mother's_occupation.csv", sep=";")

df.columns = df.columns.str.strip()

df["Father's occupation"] = df["Father's occupation"].apply(
    lambda x: "Unskilled Workers" if x == 9
    else "Skilled Workers in Industry, Construction and Craftsmen" if x == 7
    else "Personal Services, Security and Safety Workers and Sellers" if x == 5
    else "Administrative staff" if x == 4
    else "Intermediate Level Technicians and Professions" if x == 3
    else "Installation and Machine Operators and Assembly Workers" if x == 8
    else "Armed Forces Professions" if x == 10
    else "Farmers and Skilled Workers in Agriculture, Fisheries and Forestry" if x == 6
    else "Specialists in Intellectual and Scientific Activities" if x == 2
    else "Representatives of the Legislative Power and Executive Bodies, Directors, Directors and Executive Managers" if x == 1
    else "Student" if x == 0
    else "Other Situation" if x == 90
    else "Other"
)

print(df["Father's occupation"].value_counts())
```

```python
df.to_csv("data_Categoriarara(Father's_occupation).csv", sep=";", index=False)
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.read_csv("data_Categoriarara(Father's_occupation).csv", sep=";")

encoder = OneHotEncoder(sparse_output=False)

fathers_occupation_encoded = encoder.fit_transform(df[["Father's occupation"]])

nomes = encoder.get_feature_names_out(["Father's occupation"])

fathers_occupation_df = pd.DataFrame(
    fathers_occupation_encoded,
    columns=nomes,
    index=df.index
)

df = pd.concat([df.drop(columns=["Father's occupation"]), fathers_occupation_df],axis=1) #Removendo a coluna antiga e juntando as novas

print(df.head())
```

```python
df.to_csv("data_OneHot_Father's_occupation.csv", sep=";", index=False)
```
