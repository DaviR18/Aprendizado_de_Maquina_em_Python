```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.read_csv("data_OneHot_Scholarship_holder.csv", sep=";")
encoder = OneHotEncoder(sparse_output=False)

International_encoded = encoder.fit_transform(df[["International"]])

nomes = encoder.get_feature_names_out(["International"])

International_df = pd.DataFrame(
    International_encoded,
    columns=nomes,
    index=df.index
)

df = pd.concat([df.drop(columns=["International"]), International_df],axis=1)
print(df.head())
```

```python
df.to_csv("data_OneHot_International.csv", sep=";", index=False)
```
