```python
import pandas as pd

df = pd.read_csv("semoutliers.csv", sep=";")

df.columns = df.columns.str.strip()

df["Marital status"] = df["Marital status"].apply(
    lambda x: "single" if x == 1
    else "married" if x == 2
    else "Other"
)

print(df["Marital status"].value_counts())
```

```python
df.to_csv("data_Categoriarara(Marital_status).csv", sep=";", index=False)
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.read_csv("data_Categoriarara(Marital_status).csv", sep=";") # ler o bd

df.columns = df.columns.str.strip()

encoder = OneHotEncoder(sparse_output=False) # cria o OneHotEncoder

maritalstatus_encoded = encoder.fit_transform(df[["Marital status"]]) # fazendo as matrizes

nomes = encoder.get_feature_names_out(["Marital status"]) # guardando os nomes das novas colunas na variavel nomes

maritalstatus_df = pd.DataFrame( #transformando o resultado em DataFrame
    maritalstatus_encoded,
    columns=nomes,
    index=df.index
)

df = pd.concat([df.drop(columns=["Marital status"]), maritalstatus_df], axis=1) #removendo a coluna antiga e juntando as novas

print(df.head()) #so pra ver se deu certo
```

```python
df.to_csv("data_OneHot_Marital_status.csv", sep=";", index=False) # salva no bd
```
