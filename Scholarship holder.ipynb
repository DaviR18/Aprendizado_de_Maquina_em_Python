```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.read_csv("data_OneHot_Gender.csv", sep=";")
encoder = OneHotEncoder(sparse_output=False)

Scholarship_holder_encoded = encoder.fit_transform(df[["Scholarship holder"]])

nomes = encoder.get_feature_names_out(["Scholarship holder"])

Scholarship_holder_df = pd.DataFrame(
    Scholarship_holder_encoded,
    columns=nomes,
    index=df.index
)

df = pd.concat([df.drop(columns=["Scholarship holder"]), Scholarship_holder_df],axis=1)
print(df.head())
```

```python
df.to_csv("data_OneHot_Scholarship_holder.csv", sep=";", index=False)
```
