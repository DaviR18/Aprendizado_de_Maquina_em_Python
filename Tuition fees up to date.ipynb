```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.read_csv("data_OneHot_Debtor.csv", sep=";")
encoder = OneHotEncoder(sparse_output=False)

Tuition_fees_up_to_date_encoded = encoder.fit_transform(df[["Tuition fees up to date"]])

nomes = encoder.get_feature_names_out(["Tuition fees up to date"])

Tuition_fees_up_to_date_df = pd.DataFrame(
    Tuition_fees_up_to_date_encoded,
    columns=nomes,
    index=df.index
)

df = pd.concat([df.drop(columns=["Tuition fees up to date"]), Tuition_fees_up_to_date_df],axis=1)
print(df.head())
```

```python
df.to_csv("data_OneHot_Tuition_fees_up_to_date.csv", sep=";", index=False)
```
